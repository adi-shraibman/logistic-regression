## On your own: who will pass the first year?

A college wants to predict whether an applicant will **pass the first year** ($y = 1$) from two
entrance exams, **math** and **English** (each scored 0 to 100). The cell below loads data on 300 past
students (`X_app`, `y_app`: training) and on another 300 students (`X_app_test`, `y_app_test`: test).

The college will admit an applicant if $h(x) \ge t$. Two kinds of mistake:

- **admitting** a student who then **fails** (a false positive) costs the college **3** units;
- **rejecting** a student who **would have passed** (a false negative) costs **1** unit.

**Your task**, all the code yours (01b has everything you need; `plot_probability_map(w, b, X, y)` from
the helpers draws a probability map):

1. Look at the training data, fit a logistic-regression model, and say which exam matters more.
2. Evaluate it on the **test** students at threshold 0.5: store the accuracy in `acc_app`.
3. Choose the threshold $t$ (from `thresholds`) with the lowest **total cost on the training data**,
   and store it in `best_t`.

```python
def make_applicants(n, seed):
    # Entrance-exam scores (math, English); y = 1 means the student passed the first year.
    rng = np.random.default_rng(seed)
    math = np.clip(rng.normal(70, 12, n), 30, 100).round()
    english = np.clip(rng.normal(70, 12, n), 30, 100).round()
    z = 0.12 * (math - 72) + 0.08 * (english - 68)                # (hidden from the learner!)
    y = (rng.uniform(size=n) < 1 / (1 + np.exp(-z))).astype(int)
    return np.column_stack([math, english]), y

X_app, y_app = make_applicants(300, seed=10)          # past students: training data
X_app_test, y_app_test = make_applicants(300, seed=11)  # a different group: test data
thresholds = np.round(np.arange(0.05, 0.96, 0.05), 2)
print("training:", X_app.shape, "  fraction who passed:", y_app.mean().round(3))
```
